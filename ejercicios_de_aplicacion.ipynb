### 8. Reto integrador: pedido con entrega 
**Consigna:** incorpora “Retiro en local” y “Entrega a domicilio” mediante dos `Radiobutton` que compartan un `StringVar`.
1. Añade un campo de dirección.
2. Exige dirección únicamente para entrega a domicilio.
3. Agrega G. 10.000 de envío cuando se elija domicilio.
4. Calcula el descuento sobre los productos y después suma el envío.
5. Muestra modalidad, dirección cuando corresponda, subtotal, descuento, envío y total final.
6. Restablece “Retiro en local” y dirección vacía al limpiar.  

**Prueba:** productos por G. 45.000, descuento del 10% y domicilio → G. 50.500. Sin descuento y con retiro → G. 45.000.

```python
import tkinter as tk
from tkinter import ttk, messagebox

PRODUCTOS = ("Alcohol 70%", "Gasas", "Guantes", "Termómetro")

def moneda(valor):
    return "G. " + f"{valor:,}".replace(",", ".")

ventana = tk.Tk()
ventana.title("Farmacia – Ficha de pedido")
ventana.geometry("700x720")
ventana.minsize(580, 560)
ventana.columnconfigure(0, weight=1)
ventana.rowconfigure(0, weight=1)

estilo = ttk.Style(ventana)
estilo.configure("Titulo.TLabel", font=("Arial", 18, "bold"))
estilo.configure("Principal.TButton", font=("Arial", 11, "bold"), padding=(12, 8))

nombre_var = tk.StringVar(ventana)
telefono_var = tk.StringVar(ventana)
producto_var = tk.StringVar(ventana)
precio_var = tk.StringVar(ventana)
cantidad_var = tk.StringVar(ventana, value="1")
resumen_var = tk.StringVar(ventana, value="Completa los datos y pulsa Calcular pedido.")
estado_var = tk.StringVar(ventana, value="Listo para registrar un pedido.")
descuento_var = tk.BooleanVar(ventana, value=False) # Ejercicio 6
modalidad_var = tk.StringVar(ventana, value="local") # Ejercicio 8
direccion_var = tk.StringVar(ventana, value="") # Ejercicio 8

principal = ttk.Frame(ventana, padding=15)
principal.grid(row=0, column=0, sticky="nsew")
principal.columnconfigure(0, weight=1)
principal.rowconfigure(5, weight=1)
ttk.Label(principal, text="Nuevo pedido", style="Titulo.TLabel").grid(row=0, column=0, sticky="w", pady=(0, 12))

cliente = ttk.LabelFrame(principal, text="1. Datos del cliente", padding=10)
cliente.grid(row=1, column=0, sticky="ew", pady=5)
cliente.columnconfigure(1, weight=1)
ttk.Label(cliente, text="Nombre:").grid(row=0, column=0, sticky="w", padx=5, pady=5)
entrada_nombre = ttk.Entry(cliente, textvariable=nombre_var)
entrada_nombre.grid(row=0, column=1, sticky="ew", padx=5, pady=5)
ttk.Label(cliente, text="Teléfono:").grid(row=1, column=0, sticky="w", padx=5, pady=5)
entrada_telefono = ttk.Entry(cliente, textvariable=telefono_var)
entrada_telefono.grid(row=1, column=1, sticky="ew", padx=5, pady=5)

producto = ttk.LabelFrame(principal, text="2. Datos del producto", padding=10)
producto.grid(row=2, column=0, sticky="ew", pady=5)
producto.columnconfigure(1, weight=1)
ttk.Label(producto, text="Producto:").grid(row=0, column=0, sticky="w", padx=5, pady=5)
combo = ttk.Combobox(producto, textvariable=producto_var, values=PRODUCTOS, state="readonly")
combo.grid(row=0, column=1, sticky="ew", padx=5, pady=5)
ttk.Label(producto, text="Precio unitario (Gs.):").grid(row=1, column=0, sticky="w", padx=5, pady=5)
entrada_precio = ttk.Entry(producto, textvariable=precio_var)
entrada_precio.grid(row=1, column=1, sticky="ew", padx=5, pady=5)
ttk.Label(producto, text="Cantidad:").grid(row=2, column=0, sticky="w", padx=5, pady=5)
entrada_cantidad = ttk.Entry(producto, textvariable=cantidad_var)
entrada_cantidad.grid(row=2, column=1, sticky="ew", padx=5, pady=5)
ttk.Label(producto, text="Ejemplo de precio: 15000, sin puntos ni comas.").grid(
    row=3, column=0, columnspan=2, sticky="w", padx=5, pady=5)
# Ejercicio 6
chk_descuento = ttk.Checkbutton(producto, text="Aplicar descuento del 10%", variable=descuento_var)
chk_descuento.grid(row=4, column=0)

def leer_pedido():
    nombre = nombre_var.get().strip()
    telefono = telefono_var.get().strip()
    if not nombre:
        raise ValueError("Completa el nombre del cliente.")
    if not (telefono.isascii() and telefono.isdigit() and 6 <= len(telefono) <= 15):
        raise ValueError("El teléfono debe contener entre 6 y 15 dígitos, sin signos ni espacios.")
    elegido = producto_var.get()
    if elegido not in PRODUCTOS:
        raise ValueError("Selecciona un producto del catálogo.")
    try:
        precio = int(precio_var.get().strip())
    except ValueError:
        raise ValueError("El precio debe ser un entero, por ejemplo 15000.")
    if precio <= 0:
        raise ValueError("El precio debe ser mayor que cero.")
    try:
        cantidad = int(cantidad_var.get().strip())
    except ValueError:
        raise ValueError("La cantidad debe ser un entero, por ejemplo 3.")
    
    # Ejercicio 5: validación con límites
    if not 1 <= cantidad <= 100:
        raise ValueError("La cantidad debe estar entre 1 y 100")

    # Ejercicio 8: validación de dirección si es a domicilio
    modalidad = modalidad_var.get().lower()
    direccion = direccion_var.get().strip()
    if modalidad == "domicilio" and not direccion:
        raise ValueError("Indica la dirección para la entrega a domicilio.")
    
    return nombre, telefono, elegido, precio, cantidad, modalidad, direccion

def calcular(event=None):
    # Quitamos el resumen anterior para que un error no deje visible un total viejo.
    resumen_var.set("Pedido pendiente de validación.")
    try:
        nombre, telefono, elegido, precio, cantidad, modalidad, direccion = leer_pedido()
    except ValueError as error:
        estado_var.set("Revisa los datos del pedido.")
        messagebox.showwarning("Datos incompletos o incorrectos", str(error), parent=ventana)
        return
    subtotal = precio * cantidad

    # Ejercicio 6
    descuento = 0
    if descuento_var.get():
        descuento = subtotal // 10
    subtotal_con_descuento = subtotal - descuento

    # Ejercicio 8
    envio = 0
    if modalidad == "domicilio":
        envio = 10000

    total = subtotal_con_descuento + envio

    texto_resumen =  (
        f"Cliente: {nombre}\nTeléfono: {telefono}\n"
        f"Modalidad de entrega: {'Retiro en local' if modalidad=='local' else f'Envío a domicilio\tDirección: {direccion}'}\n"
        f"Producto: {elegido}\n"
        f"Precio unitario: {moneda(precio)}\nCantidad: {cantidad}\n"
        f"Subtotal: {moneda(subtotal)}\n"
        f"Descuento: {moneda(descuento)}\n"
        f"Envío: {moneda(envio)}\n"
        f"Total: {moneda(total)}"
    )

    resumen_var.set(texto_resumen)
    estado_var.set("Cálculo correcto. El pedido todavía no se guarda en un archivo.")
    

def limpiar():
    # Ejercicio 7
    confirmar = messagebox.askyesno(title="Confirmar limpieza", message="¿Deseas limpiar el formulario?", icon="question")
    if not confirmar: return

    nombre_var.set("")
    telefono_var.set("")
    producto_var.set("")
    precio_var.set("")
    cantidad_var.set("1")
    resumen_var.set("Completa los datos y pulsa Calcular pedido.")
    estado_var.set("Formulario limpio.")
    entrada_nombre.focus_set()
    # Ejercicio 6
    descuento_var.set(False)
    # Ejercicio 8
    modalidad_var.set("local")
    direccion_var.set("")

# Ejercicio 8
modalidad = ttk.LabelFrame(principal, text="3. Modalidad de entrega", padding=10)
modalidad.grid(row=3, column=0, sticky="ew", pady=5)
modalidad.columnconfigure(0, weight=1)
ttk.Radiobutton(modalidad, text="Retiro en local", value="local", variable=modalidad_var).grid(row=0, column=0, sticky="ew")
ttk.Radiobutton(modalidad, text="Envío a domicilio", value="domicilio", variable=modalidad_var).grid(row=1, column=0, sticky="ew")
direccion = ttk.Frame(modalidad)
direccion.grid(row=3, column=0, sticky="ew")
direccion.columnconfigure(1, weight=1)
ttk.Label(direccion, text="Dirección (solo para envío a domicilio): ").grid(row=0, column=0, sticky="w")
ttk.Entry(direccion, textvariable=direccion_var).grid(row=0, column=1, sticky="ew")

acciones = ttk.Frame(principal)
acciones.grid(row=4, column=0, sticky="ew", pady=10)
ttk.Button(acciones, text="Calcular pedido", command=calcular, style="Principal.TButton").grid(row=0, column=0, padx=(0, 8))
ttk.Button(acciones, text="Limpiar", command=limpiar).grid(row=0, column=1)

resumen = ttk.LabelFrame(principal, text="3. Resumen", padding=10)
resumen.grid(row=5, column=0, sticky="nsew", pady=5)
resumen.columnconfigure(0, weight=1)
resumen.rowconfigure(0, weight=1)
etiqueta_resumen = ttk.Label(resumen, textvariable=resumen_var, anchor="nw", justify="left", wraplength=500)
etiqueta_resumen.grid(row=0, column=0, sticky="nsew")
etiqueta_resumen.bind("<Configure>", lambda evento: etiqueta_resumen.configure(wraplength=max(100, evento.width)))
ttk.Label(principal, textvariable=estado_var, wraplength=530).grid(row=6, column=0, sticky="w", pady=(5, 0))

ventana.bind("<Return>", calcular)
entrada_nombre.focus_set()
ventana.mainloop()
```

### Explicaciones para la entrega

#### 1. Modificación de distribución
Agregué un contenedor (`LabelFrame`) llamado **"3. Modalidad de entrega"** y lo organicé con el gestor de geometría `grid` y `sticky="ew"` para que se adapte al ancho de la ventana. Dentro del marco, añadí los botones de opción (`Radiobutton`) para elegir entre retiro y envío, además de un `Frame` interno con el campo de dirección (`Entry`). Configuré la columna de la entrada con `columnconfigure(1, weight=1)` y `sticky="ew"` para que el campo se ensanche al ampliar la ventana y la interfaz mantenga una distribución ordenada.

#### 2. Regla de validación
Se implementaron dos validaciones clave dentro de la función `leer_pedido()`:
* **Límite de cantidad (`not (1 <= cantidad <= 100)`):** Se utiliza el operador `not` antes del intervalo para identificar las cantidades que están **fuera** de los límites permitidos. Sin `not`, la condición identificaría como error los valores permitidos, como 1 o 50, en lugar de los incorrectos, como 0 o 101. Así, el `ValueError` solo se activa cuando se ingresa un número menor que 1 o mayor que 100.
* **Validación condicional de dirección:** Se verifica si el usuario seleccionó "Entrega a domicilio". En ese caso, se comprueba que el campo de dirección no esté vacío (`not direccion`). Si está vacío, se interrumpe el cálculo y se muestra un mensaje que solicita la dirección.

### Preguntas de comprensión
1. ¿Qué diferencia hay entre `Frame` y `LabelFrame`?  
La diferencia es que Frame es un contenedor genérico e invisible utilizado únicamente para agrupar y organizar widgets y así ordenar la interfaz. Por otro lado, LabelFrame es similar al Frame pero incluye una delimitación con un borde y un título en la parte superior, utilizado para agrupar campos relacionados  

2. ¿Qué determina el padre de un widget?  
El padre de un widget determina en donde vive el widget, dibujándose dentro del contenedor padre que se indica como primer argumento (ej: `ttk.Label(principal, ...)`). Además, si define su comportamiento al cerrar, si el contenedor padre se destruye o se oculta, todos sus hijos se eliminar con él
3. ¿Por qué `sticky="ew"` puede no ser suficiente para ampliar una entrada?  
Porque sticky="ew" solo indica el widget que se estire horizontalmente dentro de su celda de la cuadrícula. Sin embargo, si la celda no está configurada para crecer (es decir, su peso por defecto es 0), la celda se mantendrá pequeña y la entrada no se expandirá, por lo que el sticky no tendría efecto. Por tal motivo, para que funcione, la columna se debe configurar con `columnconfigure(..., weight=1)`
4. ¿Qué diferencia existe entre `padding` y `padx`/`pady`?  
La diferencia es que padding es un espaciado interno y se aplica al configurar el contenedor o widget para dejar espacio entre su borde y los elementos que contiene. Por otro lado, padx/pady son espaciados externos o márgenes, y se definen al usar el gestor de geometría (.grid()) para separar un widget específico de otros widgets vecinos o de los bordes de la celda
5. ¿Por qué comprobamos que una cantidad sea positiva después de convertirla a entero?  
Porque la función de conversión int() solo transforma el texto ingresado en un número entero, pero no valida el signo de ese número. Un usuario podría escribir un número negativo (como -5), que técnicamente es un entero válido pero es lógicamente incorrecto para una cantidad, precio o stock  
6. ¿Cómo evita el programa mostrar un total anterior después de un error?  
Se evita limpiando explícitamente la variable o etiqueta de resultado dentro del bloque de manejo de errores (except). Si ocurre un fallo (por ejemplo, al intentar convertir texto inválido), el bloque de error asigna un texto vacío ("") o un mensaje de error a la etiqueta de salida, haciendo que se sobrescriba o borre cualquier resultado viejo que había en pantalla
7. ¿Por qué `calcular(event=None)` funciona con el botón y con Enter?  
Funciona gracias a la combinación del parámetro por defecto `event=None` y el modelo de eventos de Tkinter. Al asociarlo a un botón (command=calcular), la función se invoca sin argumentos y event toma por defecto el valor None. En cambio, al presionar Enter, la línea `ventana.bind("<Return>", calcular)` hace que Tkinter envíe automáticamente un objeto de tipo Event que reemplaza a ese None. Como el código interno no necesita consultar los datos de la pulsación, la función procesa ambas llamadas de manera totalmente transparente y sin errores
